Task 06:

```python
# Task 06 code
regions = np.array(["Punjab", "Sindh", "KP", "Balochistan", "Islamabad", "AJK"])
p17 = np.array([110.01, 47.85, 35.53, 12.34, 2.01, 4.05])
census23 = np.array([127.69, 55.70, 40.85, 14.89, 2.36, 4.05])

def pairsDict(dicePairs):
    return {str(kar): (v.item() if hasattr(v, "item") else v) for kar, v in dicePairs}

total = np.sum(census23)
print(f"Ex6a => total population 2023 = {total:.2f} million")

pctShare = 100 * census23 / total
print(f"Ex6b => share of total (%): {pairsDict(zip(regions, np.round(pctShare, 2)))}")

print(f"Ex6c => most populous unit: {regions[census23.argmax()]}")
print(f"Ex6d => units above 20 million: {regions[census23 > 20]}")

annGrowth = (np.power(census23 / p17, 1 / 6) - 1) * 100
print(f"Ex6e => annual growth 2017-2023 (% per year): {pairsDict(zip(regions, np.round(annGrowth, 2)))}")

print(f"Ex6f => fastest-growing unit: {regions[np.argmax(annGrowth)]} ({annGrowth.max():.2f}% per year)")

natRate = ((census23.sum() / p17.sum()) ** (1 / 6) - 1) * 100
dblTime = 70 / annGrowth.max()
print(f"Ex6 interpret => national growth = {natRate:.2f}% per year.")
print(f"Ex6 interpret => {regions[np.argmax(annGrowth)]} would double in about {dblTime:.0f} years at its current rate.")
print(f"Ex6 interpret => AJK grew {annGrowth[-1]:.2f}% per year vs national {natRate:.2f}% -> "
      f"AJK is {natRate - annGrowth[-1]:.2f} points below the national rate (in this sample AJK is basically flat).")
```

Task 07:

```python
# Task 07 code
cities = np.array(["Lahore", "Karachi", "Islamabad", "Quetta", "Peshawar"])
monthNames = np.array(["Jan", "Feb", "Mar", "Apr", "May", "Jun",
                   "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])

temperature = np.array([
    [19, 22, 28, 35, 40, 40, 36, 34, 34, 32, 26, 21],
    [26, 28, 32, 34, 35, 34, 32, 31, 32, 34, 31, 28],
    [17, 19, 24, 30, 36, 38, 35, 33, 33, 30, 24, 19],
    [11, 13, 18, 24, 29, 34, 35, 34, 31, 25, 19, 13],
    [18, 20, 25, 31, 37, 40, 38, 36, 35, 31, 25, 19],
])
precip = np.array([
    [12, 22, 34, 22, 21, 54, 214, 232, 82, 8, 3, 8],
    [8, 10, 9, 4, 1, 6, 40, 25, 13, 1, 2, 5],
    [56, 65, 76, 52, 32, 66, 260, 310, 120, 28, 12, 28],
    [38, 40, 33, 20, 8, 5, 18, 15, 5, 3, 7, 18],
    [33, 48, 63, 48, 23, 21, 86, 95, 43, 23, 20, 29],
])

def pairsDict(dicePairs):
    return {str(kar): (v.item() if hasattr(v, "item") else v) for kar, v in dicePairs}

print(f"Ex7a => shape of temp (cities, months): {temperature.shape}")

avgTemp = np.mean(temperature, axis=1)
print(f"Ex7b => yearly mean temperature (deg C): {pairsDict(zip(cities, np.round(avgTemp, 1)))}")

hottest = monthNames[np.argmax(temperature, axis=1)]
print(f"Ex7c => hottest month per city: {pairsDict(zip(cities, hottest))}")

print(f"Ex7d => 5-city monthly average (deg C): {pairsDict(zip(monthNames, np.round(temperature.mean(axis=0), 1)))}")

over35 = np.count_nonzero(temperature > 35, axis=1)
print(f"Ex7e => months above 35 deg C: {pairsDict(zip(cities, over35))}")

lhrF = temperature[0] * 9 / 5 + 32
print(f"Ex7f => Lahore monthly temperature (deg F): {np.round(lhrF, 1)}")

rainJas = precip[:, 6:9].sum(1)
print(f"Ex7g => monsoon rain Jul-Sep (mm): {pairsDict(zip(cities, rainJas))}")
print(f"Ex7g => wettest monsoon city: {cities[np.argmax(rainJas)]} ({rainJas.max()} mm)")

pctShare = rainJas / precip.sum(axis=1) * 100
print(f"Ex7h => monsoon share of annual rain (%): {pairsDict(zip(cities, np.round(pctShare, 1)))}")

kar = np.where(cities == "Karachi")[0][0]
print(f"Ex7 interpret => Karachi has the highest yearly mean ({avgTemp.max():.1f} deg C) "
      f"but {over35[kar]} months above 35 deg C, while cooler-on-average Peshawar has {over35[4]} "
      f"and Lahore {over35[0]}. A high yearly mean does not mean the most extreme heat: "
      f"Karachi is hot all year, the inland cities spike in summer.")
```

Task 08:

```python
# Task 08 code
fiscalYears = np.array(["FY2023", "FY2024", "FY2025"])
fyMonths = np.array(["Jul", "Aug", "Sep", "Oct", "Nov", "Dec",
                    "Jan", "Feb", "Mar", "Apr", "May", "Jun"])

inflData = np.array([
    [24.9, 27.3, 23.2, 26.6, 23.8, 24.5, 27.6, 31.5, 35.4, 36.4, 38.0, 29.4],
    [28.3, 27.4, 31.4, 26.9, 29.2, 29.7, 28.3, 23.1, 20.7, 17.3, 11.8, 12.6],
    [4.1, 9.6, 6.9, 7.2, 4.9, 4.1, 2.4, 1.5, 0.7, 0.3, 3.5, 3.2],
])

def pairsDict(dicePairs):
    return {str(kar): (v.item() if hasattr(v, "item") else v) for kar, v in dicePairs}

avgRate = np.mean(inflData, axis=1)
print(f"Ex8a => average inflation per year (%): {pairsDict(zip(fiscalYears, np.round(avgRate, 2)))}")

fy24 = inflData[1]
print(f"Ex8b => FY2024 months above 25 %: {fyMonths[fy24 > 25]} -> {(fy24 > 25).sum()} months")

delta = np.diff(inflData, axis=1)
print(f"Ex8c => month-on-month change (percentage points; negative = drop):\n{np.round(delta, 1)}")

dropIdx = delta.argmin(axis=1)
print(f"Ex8d => month of biggest drop: {pairsDict(zip(fiscalYears, fyMonths[dropIdx + 1]))} | size (pp): {np.round(delta.min(axis=1), 1)}")

inflFy25 = inflData[2]
band = np.where(inflFy25 < 5, "Low", np.where(inflFy25 < 10, "Moderate", "High"))
print(f"Ex8e => FY2025 labels (Low < 5 %, Moderate 5-10 %, High >= 10 %):\n{np.column_stack((fyMonths, inflFy25, band))}")

print(f"Ex8 interpret => average inflation went {avgRate[0]:.1f}% -> {avgRate[1]:.1f}% -> {avgRate[2]:.1f}%. "
      f"Prices did NOT fall; they kept rising, but more and more slowly (disinflation), "
      f"so the price LEVEL is still much higher than at the start of FY2023.")
```

Task 09:

```python
# Task 09 code
marks = np.array([34, 38, 35, 52, 67, 73, 88, 95, 98, 39,
                  41, 45, 36, 58, 62, 71, 80, 33, 99, 44])
studentRolls = np.arange(1, 21)
CUTOFF = 40

scoresPlus5 = np.clip(5 + marks, 0, 100)
print(f"Ex9a => original marks: {marks}")
print(f"Ex9a => marks after +5 grace (cap 100): {scoresPlus5}")

beforePct = (marks >= CUTOFF).mean() * 100
after = 100 * np.mean(scoresPlus5 >= CUTOFF)
print(f"Ex9b => pass % before grace = {beforePct:.1f} %, after grace = {after:.1f} %")

graceOnly = (marks < CUTOFF) & (scoresPlus5 >= CUTOFF)
print(f"Ex9c => passed only because of grace (roll numbers): {studentRolls[graceOnly]}")
print(f"Ex9c => their original marks: {marks[graceOnly]}")

zVals = (scoresPlus5 - np.mean(scoresPlus5)) / np.std(scoresPlus5)
print(f"Ex9d => mean = {scoresPlus5.mean():.2f}, std = {scoresPlus5.std():.2f}")
print(f"Ex9d => z-scores (std devs from mean): {np.round(zVals, 2)}")
```

```python
# Task 10 code
digitImg = np.zeros((8, 8), dtype=np.uint8)
digitImg[1, 1:-1] = 255
digitImg[[2, 3, 4, 5, 6], [6, 5, 5, 4, 4]] = 255

def binarize(a):
    return (a > 0).astype(int)

print(f"Ex10a => the '7' (1 = white):\n{binarize(digitImg)}")

flipped = np.fliplr(digitImg)
print(f"Ex10b => flipped left-right:\n{binarize(flipped)}")

negative = ~digitImg
print(f"Ex10c => inverted (white <-> black):\n{binarize(negative)}")

imgCrop = digitImg[0:7, 1:8]
print(f"Ex10d => cropped 7x7 region, shape: {imgCrop.shape}\n{binarize(imgCrop)}")

brighter = np.clip(digitImg.astype(int) + 60, 0, 255).astype(np.uint8)
print(f"Ex10e => brightened (+60, capped at 255); unique values: {np.unique(brighter)}")

print(f"Ex10f => shape = {digitImg.shape} | dtype = {digitImg.dtype} | white pixels = {(digitImg == 255).sum()}")
```
